# Ejecucion

## Importar librerias y cargar datos

```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

```python
# Mostrar las primeras 5 filas
netflix_df.head()
```

## Datos de la decada de 1990

```python
# Filtrar los datos de Netflix en la década de 1990
netflix_1990 = netflix_df[(netflix_df['release_year'] >= 1990) & (netflix_df['release_year'] < 2000)]
netflix_1990
```

## Encontrar las peliculas con mayor duracion

```python
# Variable con la duracion mas frecuente
duration = netflix_1990['duration'].mode()[0] # Se usa el metodo estadistico moda
print(f'La duracion mas comun fue: {duration} min')
```

```python
# Establecer un histograma de la duracion
plt.figure(figsize=(10, 6))
plt.hist(netflix_1990['duration'], bins=20)
plt.title('Distribución de las Duraciones de Películas en la Década de 1990')
plt.xlabel('Duración (minutos)')
plt.ylabel('Frecuencia')
plt.axvline(duration, color='r', linestyle='dashed', linewidth=1)
plt.text(duration + 5, plt.ylim()[1] * 0.9, f'Moda: {duration} min', color='r')
plt.show()
```

```python
# Estableciendo la cuenta de peliculas de accion mas corta
action_movies_shorts =  netflix_1990[(netflix_1990['genre'] == 'Action') & (netflix_1990['duration'] < 90)]
action_movies_shorts
```

```python
# Conteo de peliculas de accion cortas
short_movie_count = len(action_movies_shorts)
print(short_movie_count)
```
