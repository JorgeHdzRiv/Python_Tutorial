## Diagrama de dispersion

Los diagramas de dispersión usan una colección de puntos colocados usando coordenadas cartesianas para mostrar valores de dos variables. Al mostrar una variable en cada eje, se puede detectar si existe una relación o correlación entre las dos variables.

```
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show()
```


![](./img/diagrama_de_dispersion.svg)

````python
# Cambiando a grafico de dispersion sin escala
import matplotlib.pyplot as plt
plt.scatter(gdp_cap,life_exp)
plt.xlabel('GDP per capita [in USD]')
plt.ylabel('Life Expectancy [in years]')
plt.title('Country GDP per Capita vs Life Expectancy')
plt.show()
````

# Histogramas

Un histograma es un gráfico que usa barras para simbolizar cómo se distribuye un conjunto de datos. 

También sirve para ver rápidamente cómo se ha comportado una muestra basada en una variable numérica o cuantitativa.

![](./img/hist.png)

````python
# Creacion de un histograma a partir de la expectativa de vida
import matplotlib.pyplot as plt
plt.figure(figsize=(14,6))
plt.hist(life_exp)
plt.title('Expectativa de Vida')
````

## Histograma 2

En el ejercicio anterior, no especificaste el número de rangos. Por defecto, en ese caso Python establece el número de rangos en 10. El número de rangos es bastante importante. Muy pocos rangos simplificarán demasiado la realidad y no te mostrarán los detalles. Demasiados rangos complicarán demasiado la realidad y no mostrarán el panorama general.

Para controlar el número de rangos en los que dividir tus datos, puedes establecer el argumento **bins**.



````python
# Construye un histograma con 5 bins
plt.hist(life_exp,bins=5)

# Muestra y limpia la gráfica
plt.show()
plt.clf()

# Construye histograma con 20 rangos
plt.hist(life_exp,bins=20)

# Muestra y limpia de nuevo
plt.show()
plt.clf()
````

**Aumentando los bins se puede ver mejor la distribucion de los datos**

**Descripcion de s:**

s: flotante o similar a una matriz, forma (n,), opcional
